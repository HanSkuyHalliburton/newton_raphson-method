```python
import numpy as np
import pandas as pd
import math

def func(x):
    return x*math.sin(x)-math.e**x
def dfunc(x):
    return math.sin(x)+x*math.cos(x)-math.e**x
def newton(x0, tol):
    rows = []
    i = 1

    while True:
        if dfunc(x0) == 0:
            print("gabisa boss")
            return None
        x1 = x0 - func(x0) / dfunc(x0)
        err = abs(x1 - x0) / abs(x1) *100 if x1 !=0 else np.inf

        rows.append([i, x0, func(x0), dfunc(x0), x1, err])
    
        if err <= tol or func(x1) == 0:
            break
        x0=x1
        i += 1
        
    df = pd.DataFrame(rows, columns=["Iterasi", "x_old", "f(x_old)", "f'(x_old)", "x_new", "Error (%)"])

    display(df.style.format({"x_old": "{:.5f}", "f(x_old)": "{:.5f}", "f'(x_old)": "{:.5f}", "x_new": "{:.5f}", "Error (%)": "{:.5f}"})
        .hide(axis="index")
        .set_caption("Tabel Hasil Pencarian Akar dengan Metode Newton-Raphson")
        .set_properties(**{"text-align": "center"})
        .set_table_styles([{"selector": "caption", "props": [("font-size", "14px"), ("font-weight", "bold"),
                                                             ("text-align", "center"), ("padding", "8px")]},
                           {"selector": "th", "props": [("background-color", "#FFF00"), ("color", "white"),
                                                        ("text-align", "center"), ("padding", "8px")]},
                           {"selector": "td", "props": [("padding", "6px 12px")]},
                           {"selector": "tbody tr:nth-child(even)", "props": [
                               ("background-color", "#E8EBEA")
                           ]}
                           ]))
    print(f"root of f(x) = {x1:.6f}")
    return x1
root = newton(6, 0.001)
```
